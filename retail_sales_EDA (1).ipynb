```python
df["month"] = df["sale_date"].dt.to_period("M")
```

```python
monthly_sales = df.groupby("month")["total_sale"].sum()

monthly_sales
# ### Observation

# The monthly sales trend shows how total sales changed over time.
# The highest-sales month indicates the strongest monthly performance,
# while the lowest-sales month represents a comparatively weaker period.

# These trends can help the business plan inventory, marketing campaigns,
# and resources according to monthly demand.
```

```python
plt.figure(figsize=(14, 5))

plt.plot(
    monthly_sales.index.astype(str),
    monthly_sales.values,
    marker="o"
)

plt.title("Monthly Sales Trend")
plt.xlabel("Month")
plt.ylabel("Total Sales")

plt.xticks(rotation=45)
plt.grid(True)

plt.tight_layout()
plt.show()
```

```python
highest_sales_month = monthly_sales.idxmax()
highest_sales_value = monthly_sales.max()

print("Highest Sales Month:", highest_sales_month)
print("Highest Sales:", highest_sales_value)
```

```python
lowest_sales_month = monthly_sales.idxmin()
lowest_sales_value = monthly_sales.min()

print("Lowest Sales Month:", lowest_sales_month)
print("Lowest Sales:", lowest_sales_value)
```

```python
df["quarter"] = df["sale_date"].dt.to_period("Q")
```

```python
quarterly_sales = df.groupby("quarter")["total_sale"].sum()

quarterly_sales
```

```python
plt.figure(figsize=(12, 5))

plt.plot(
    quarterly_sales.index.astype(str),
    quarterly_sales.values,
    marker="o"
)

plt.title("Quarterly Sales Trend")
plt.xlabel("Quarter")
plt.ylabel("Total Sales")

plt.grid(True)
plt.tight_layout()
plt.show()
### Observation

#The quarterly sales trend provides a broader view of business performance
#than monthly analysis. It helps identify the strongest and weakest quarters
#and can support inventory planning, marketing decisions, and resource
#allocation.
```

```python
highest_quarter = quarterly_sales.idxmax()
highest_quarter_sales = quarterly_sales.max()

print("Highest Sales Quarter:", highest_quarter)
print("Highest Sales:", highest_quarter_sales)
```

```python
lowest_quarter = quarterly_sales.idxmin()
lowest_quarter_sales = quarterly_sales.min()

print("Lowest Sales Quarter:", lowest_quarter)
print("Lowest Sales:", lowest_quarter_sales)
```

```python
bins = [0, 18, 25, 35, 45, 55, 65, 100]

labels = [
    "<18",
    "18-25",
    "26-35",
    "36-45",
    "46-55",
    "56-65",
    "65+"
]

df["age_group"] = pd.cut(
    df["age"],
    bins=bins,
    labels=labels
)
```

```python
age_group_counts = df["age_group"].value_counts().sort_index()

age_group_counts
```

```python
plt.figure(figsize=(10, 5))

age_group_counts.plot(kind="bar")

plt.title("Customer Distribution by Age Group")
plt.xlabel("Age Group")
plt.ylabel("Number of Customers")

plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
### Observation

#The age-group distribution shows how transactions are distributed across
#different customer age groups. The largest age group represents an
#important customer segment that can be targeted through focused marketing
#campaigns and personalized offers.
```

```python
highest_age_group = age_group_counts.idxmax()
highest_age_count = age_group_counts.max()

print("Highest Age Group:", highest_age_group)
print("Number of Customers:", highest_age_count)
```

```python
gender_counts = df["gender"].value_counts()

gender_counts
```

```python
plt.figure(figsize=(7, 5))

gender_counts.plot(kind="bar")

plt.title("Customer Distribution by Gender")
plt.xlabel("Gender")
plt.ylabel("Number of Customers")

plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
### Observation

#The gender distribution shows the contribution of different customer
#groups to total transactions. Understanding this distribution can help
#the business design more targeted marketing campaigns and customer offers.
```

```python
gender_percentage = (
    df["gender"]
    .value_counts(normalize=True)
    .mul(100)
    .round(2)
)

gender_percentage
```

```python
df["category"].value_counts()
category_quantity = (
    df.groupby("category")["quantity"]
      .sum()
      .sort_values(ascending=False)
)

category_quantity
plt.figure(figsize=(8, 5))

category_quantity.plot(kind="bar")

plt.title("Sales Volume by Product Category")
plt.xlabel("Product Category")
plt.ylabel("Total Quantity Sold")

plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
### Observation

#The category-level analysis shows differences in sales volume and revenue
#across product categories. The highest-volume category indicates strong
#customer demand, while the highest-revenue category contributes the most
#to overall sales value.

#Comparing quantity sold with revenue can help the business identify
#high-volume and high-value categories.
```

```python
highest_category = category_quantity.idxmax()
highest_quantity = category_quantity.max()

print("Highest Selling Category:", highest_category)
print("Quantity Sold:", highest_quantity)
```

```python
category_revenue = (
    df.groupby("category")["total_sale"]
      .sum()
      .sort_values(ascending=False)
)

category_revenue
```

```python
plt.figure(figsize=(8, 5))

category_revenue.plot(kind="bar")

plt.title("Revenue by Product Category")
plt.xlabel("Product Category")
plt.ylabel("Total Revenue")

plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
highest_revenue_category = category_revenue.idxmax()
highest_revenue = category_revenue.max()

print("Highest Revenue Category:", highest_revenue_category)
print("Total Revenue:", highest_revenue)
### Data Limitation

#The dataset does not contain individual product names or product IDs.
#Therefore, a Top 10 individual product analysis cannot be performed.
#Instead, product categories have been analyzed using total quantity sold
#and total revenue.
```

```python
import pandas as pd
df = pd.read_csv("/content/Retail_Sales.csv")

category_gender_sales = pd.pivot_table(
    df,
    values="total_sale",
    index="category",
    columns="gender",
    aggfunc="sum"
)

category_gender_sales
```

```python
import matplotlib.pyplot as plt

category_gender_sales.plot(
    kind="bar",
    figsize=(10, 5)
)

plt.title("Revenue by Product Category and Gender")
plt.xlabel("Product Category")
plt.ylabel("Total Revenue")

plt.xticks(rotation=0)
plt.legend(title="Gender")

plt.tight_layout()
plt.show()
```

```python
category_gender_sales.plot(
    kind="bar",
    figsize=(10, 5)
)

plt.title("Revenue by Product Category and Gender")
plt.xlabel("Product Category")
plt.ylabel("Total Revenue")

plt.xticks(rotation=0)
plt.legend(title="Gender")

plt.tight_layout()
plt.show()
```

```python
highest_combination = category_gender_sales.stack().idxmax()
highest_combination_value = category_gender_sales.stack().max()

print("Highest Revenue Combination:", highest_combination)
print("Revenue:", highest_combination_value)
```

### Observation

The visualization compares revenue across product categories and gender
groups. It helps identify whether certain categories generate relatively
higher revenue from a particular gender segment.

This insight can support more targeted marketing campaigns and promotional
strategies for specific customer segments.

# Actionable Business Recommendations

### Recommendation 1 — Target High-Value Customer Segments
The business should focus marketing campaigns on the customer age groups
and gender segments that contribute the most transactions and revenue.
Personalized offers can be used to improve customer engagement.

### Recommendation 2 — Prioritize High-Revenue Categories
The business should maintain adequate inventory and promotional activity
for categories that generate the highest revenue. This can help prevent
stock shortages and maximize sales opportunities.

### Recommendation 3 — Use Seasonal Sales Planning
Monthly and quarterly sales trends should be used to plan inventory,
staffing, and promotional campaigns. Resources can be increased before
strong sales periods and targeted discounts can be used during weaker
periods.

The current dataset contains product categories but does not contain
individual product names or IDs. Recording product-level information in
future transactions would allow detailed Top-10 product analysis and
better product-level recommendations.
